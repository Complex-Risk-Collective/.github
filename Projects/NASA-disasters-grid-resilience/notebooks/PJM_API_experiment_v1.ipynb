```python
"""
Authentication utilities for the PJM Data Miner API.
"""

import requests

TOKEN_URL = "https://api.pjm.com/api/v1/token"


def get_token(subscription_key: str) -> str:
    """
    Retrieve a bearer token from the PJM Data Miner API.

    Parameters
    ----------
    subscription_key : str
        Your PJM API subscription key.

    Returns
    -------
    str
        JWT bearer token.
    """

    headers = {
        "Ocp-Apim-Subscription-Key": subscription_key
    }

    response = requests.post(TOKEN_URL, headers=headers)

    response.raise_for_status()

    token = response.json()["access_token"]

    return token
```
